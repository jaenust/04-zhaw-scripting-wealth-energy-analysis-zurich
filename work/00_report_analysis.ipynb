# 00_report_analysis

# **Einleitung**

Der Kanton Zürich steht im Spannungsfeld zwischen Wohlstand und Nachhaltigkeit. Viele Gemeinden weisen ein überdurchschnittlich hohes steuerbares Einkommen bzw. Vermögen pro Einwohner auf. Diese grossen wirtschaftlichen Unterschiede werfen eine zentrale Frage auf:

> **Beeinflusst das Vermögen bzw. die wirtschaftliche Leistungsfähigkeit einer Gemeinde den Energieverbrauch ihrer Einwohner?**


Für Politik, Energieversorger und Bevölkerung ist diese Fragestellung relevant:

* **Politik:** Sollten Förderprogramme gezielt auf besonders energieintensive Gemeinden ausgerichtet werden?
* **Energieversorger:** Welche Gemeinden weisen überdurchschnittliche Lastprofile auf?
* **Bevölkerung:** Wie nachhaltig ist das eigene Energieverhalten im kantonalen Vergleich?

Ziel dieses Projekts ist es daher:

1. Zu analysieren, ob wohlhabendere Gemeinden (vermögensstark, hohes steuerbares Einkommen) tendenziell mehr oder weniger Energie pro Kopf verbrauchen.
2. Eine geographische Visualisierung (Heatmap) des Energieverbrauchs im Kanton Zürich zu erstellen, um energieintensive Regionen sichtbar zu machen.
3. Korrelationen und mögliche Zusammenhänge zwischen Vermögen und Energieverbrauch zu untersuchen.

---

# **Daten**

Für dieses Projekt wurden mehrere Datensätze zusammengeführt, welche unterschiedliche Aspekte der Fragestellung abdecken. Im Folgenden werden sie beschrieben.


## **1. Energieverbrauch pro Gemeinde (BFE / EnergieReporter)**

**Datei:** `energyreporter_municipality_historized.csv`
**Quelle:**
[https://opendata.swiss/de/dataset/energie-reporter](https://opendata.swiss/de/dataset/energie-reporter)

Dieser Datensatz enthält historische Verbrauchsdaten auf Ebene der Gemeinden.
Im Projekt wurden vor allem folgende Informationen genutzt:

* Gesamtstromverbrauch pro Gemeinde
* Stromverbrauch pro Endverbraucherkategorie (Haushalte, Dienstleistungen, Industrie etc.)
* Verwendet wurde das Jahr 2024, da dies der einzige vollständig verfügbare Datensatz war.


---

## **2. Gemeindevermögen / Steuerdaten (Statistik Kanton Zürich)**

**Datei:** `data_7354970.csv`
**Quelle:** [https://www.web.statistik.zh.ch/gpv2/#](https://www.web.statistik.zh.ch/gpv2/#)

Dieser Datensatz liefert:

* Durchschnittliches steuerbares Einkommen pro Einwohner
* Durchschnittliches steuerbares Vermögen pro Einwohner

Hier stand jedoch nur der neueste verfügbare Datensatz von 2022 zur Verfügung.

**Hinweis zur Jahrendifferenz**

Da die Steuerdaten aus 2022 stammen und die Energieverbrauchsdaten aus 2024, werden in der Analyse Werte aus zwei verschiedenen Jahren verglichen.
Diese Kombination war notwendig, da für beide Themen keine identischen Jahre vollständig vorlagen.
Für die Fragestellung ist dies jedoch ausreichend, da sich Vermögensstrukturen von Gemeinden üblicherweise nur langsam verändern.

---

## **3. Geodaten des Kantons Zürich**

**Datei:** `zh-municipalities.geojson`

Dieser Datensatz enthält:

* räumliche Abgrenzungen der Gemeinden
* BFS-Nummern zur sauberen Verknüpfung mit den Energie- und Steuerdaten

Er wird genutzt, um in *03_data_visualization* eine **Heatmap des Stromverbrauchs pro Kopf** zu erstellen.

Wir hatten hier die Herausforderung, dass es in den vergangenen Jahren zu Gemeindefusionen gekommen ist. Diese mussten wir ausfindig machen und dementsprechend die BFS Nummern aktualisieren.
